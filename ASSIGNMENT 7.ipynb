```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Load data
df = pd.read_csv(r"C:\Users\Admin\Downloads\data.csv")

# 2. Convert the flat columns into a 2D grid matrix
# This maps 'Item' to rows, 'Year' to columns, and fills cells with 'Quantity'
pivot_df = df.pivot_table(index='Item', columns='Year', values='Quantity', aggfunc='sum')

# 3. Plot the heatmap using the reshaped 2D matrix
plt.figure(figsize=(10, 6))
sns.heatmap(pivot_df, annot=True, cmap='viridis', fmt='g')

# 4. Correct Matplotlib functions for titles and display
plt.title("Item vs Year (Quantity)")
plt.xlabel("Year")
plt.ylabel("Item")

plt.show()
```
